```python
import kagglehub
import shutil
import os
from functools import reduce
import boto3
from botocore.exceptions import NoCredentialsError
```

```python
bases = [
    {'url': 'datahackers/state-of-data-brazil-2025-2026', 'ano': 2025},
    {'url': 'datahackers/state-of-data-brazil-2023', 'ano': 2023},
    {'url': 'datahackers/state-of-data-2022', 'ano': 2022},
    {'url': 'datahackers/state-of-data-brazil-20242025', 'ano': 2024},
    {'url': 'datahackers/state-of-data-2021', 'ano': 2021}
]
```

```python
AWS_REGION = 'us-east-1'
BUCKET_NAME = 'pos-tech-state-of-data-grupo-72'
```

```python
s3 = boto3.client('s3')
```

```python
for i in bases:
    p = kagglehub.dataset_download(i['url'])
    try:
        arquivo = os.listdir(p)[0]
        caminho_origem = os.path.join(p, arquivo)

        chave_s3 = f"bronze/state-of-data/{i['ano']}/{arquivo}"
        s3.upload_file(caminho_origem, BUCKET_NAME, chave_s3)

        print(f"{arquivo} enviado para s3://{BUCKET_NAME}/{chave_s3}")

    except Exception as e:
        print(f"Erro no ano {i['ano']}: {e}")
```
