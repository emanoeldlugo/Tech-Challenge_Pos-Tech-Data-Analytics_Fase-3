```python
AWS_REGION = 'us-east-1'
BUCKET_NAME = 'pos-tech-state-of-data-grupo-72'
```

```python
FOLDERS = [
    'bronze/',
    'silver/',
    'gold/',
    'athena-logs/'
]
```

```python
print('Validando credenciais  de AWS...')
!aws sts get-caller-identity
```

```python
print(f'\nCriando bucket: s3://{BUCKET_NAME}')
!aws s3 mb s3://{BUCKET_NAME} --region {AWS_REGION}
```

```python
print(f'\nCriando pastas no s3')

for folder in FOLDERS:
    print(f'Criando Pasta: {folder}')
    !aws s3api put-object --bucket {BUCKET_NAME} --key {folder}
```

```python
print(f'\nListando  estrutura do bucket s3://{BUCKET_NAME}')
!aws s3 ls s3://{BUCKET_NAME} --recursive
```
